## JAX

JAX is a library for array computation, with a NumPy-like API (`jax.numpy`) that runs on CPU, GPU and TPU. Its core features are function transformations: automatic differentiation (`jax.grad`), compilation with XLA (`jax.jit`) and automatic vectorization (`jax.vmap`). Neural networks are usually built with libraries of the JAX ecosystem, such as [Flax](https://flax.readthedocs.io/) (layers and models) and [Optax](https://optax.readthedocs.io/) (optimizers).

- [JAX Documentation](https://docs.jax.dev/)
- [JAX Quickstart](https://docs.jax.dev/en/latest/quickstart.html)
- [JAX - The Sharp Bits](https://docs.jax.dev/en/latest/notebooks/Common_Gotchas_in_JAX.html)
- [Flax Documentation](https://flax.readthedocs.io/)
